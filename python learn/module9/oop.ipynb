```python
#9.3 Inheritance
#attribute and method inheritance
#class attribute and instance attribute
class Phone:#class name should be always capital letter
    category="electronics"

#constructor method
    def __init__(self,model,price,camera,battery=100):
        self.model=model
        self.price=price
        self.camera=camera
        self.battery=battery

    def charge(self,hours):
        print(f"the phone is charging for {hours} hours")

    def photo(self,photo):
        if(self.battery>0):
            print(f"the photo is taken by {self.model} ")
            self.battery-=photo
        else:
            print(f"No battery life left in{self.model}")

```

```python
#creating object
apple = Phone("Iphone 15 Pro Max", 70000, "50mp")#object should be smaller latter
# apple.category="smart Phone"

motorola=Phone("Motorola G5",50000,"40mp")
samsung=Phone("Samsung S20",60000,"45mp")

print(type(apple),apple.category)
# print(apple.model,apple.price,apple.battery,apple.camera)
motorola.photo(10)
motorola.battery


```

```python
#9.4 oop pillars 
#inheritance,polymorphism,encapsulation,abstraction
#Inheritance: Inheritance is a mechanism in object-oriented programming that allows a class (called a child or subclass) to inherit properties and behaviors (attributes and methods) from another class (called a parent or superclass). This promotes code reusability and establishes a hierarchical relationship between classes.

#we will use previous class phone
class Smartphone(Phone):#child class
    def __init__(self,model,battery,camera,processor):
        super().__init__(model,battery,camera)
        self.processor=processor

    
```
