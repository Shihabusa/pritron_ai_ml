```python
#9.3 Inheritance
#attribute and method inheritance
#class attribute and instance attribute
class Phone:#class name should be always capital letter
    category="electronics"

#constructor method
    def __init__(self,model,price,camera,battery=100):
        self.model=model
        self.price=price
        self.camera=camera
        self.battery=battery

    def charge(self,hours):
        self.charge+=hours

    def photo(self,photo):
        if(self.battery>0):
            print(f"the photo is taken by {self.model} ")
            self.battery-=photo
        else:
            print(f"No battery life left in{self.model}")

```
