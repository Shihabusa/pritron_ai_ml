```python
#operators and expressions
#arithmetic operators
a=10
b=5
print(a+b)#+ is addition operator addition of a and b
print(a-b)#- is subtraction operator subtraction of a and b
print(a*b)#* is multiplication operator multiplication of a and b
print(a/b)#/ is division operator division of a and b
print(a%b)  #% is modulus operator 
print(a**b) #** is exponentiation operator
print(a//b) #// is floor division operator

#comparison operators
c=10
d=3
is_greater=c>d
print(is_greater) #True
is_less=c<d
print(is_less) #False
is_equal=c==d
print(is_equal) #False
is_not_equal=c!=d
print(is_not_equal) #True
is_greater_or_equal=c>=d
print(is_greater_or_equal) #True
is_less_or_equal=c<=d
print(is_less_or_equal) #False
```

```python
#logical operators
x=10;y=5;z=3
#and operator
result_and=(x>y) and (y>z)
print(result_and) #True
result_and2=(x>y) and (y<z)
print(result_and2) #False

#or operator
result_or=(x>y) or (y<z)
print(result_or) #True
#not operator
result_not=not (x>y)
print(result_not) #False
```

```python
#presidence and associativity
#arithmetic operators have higher precedence than comparison operators
#() has the highest precedence
#** has the next highest precedence
#*, /, %, // have the same precedence and are left-associative
#+, - have the same precedence and are left-associative
#comparison operators have lower precedence than arithmetic operators

a=10*2+3*2-5
print(a) #21

```
