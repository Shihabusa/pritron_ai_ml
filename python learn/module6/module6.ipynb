```python
#6.4 positional argument 
# argument 2 type (positional,keyword)

def position(a,b,c):
    print(f"first{a}, 2nd {b} and then {c}")

position(4,7,9) # jei position e je oi i print hbe

#args way to print multiple variable
def adders(*args):
    #print(args)#as a touple print hbe
    for i in args:
        print(i)

adders(10,30,40,50)
```

```python
#6.5 keywords arguments
def position(a,b,c):
    print(f"first{a}, 2nd {b} and then {c}")

position(a=4,c=7,b=9) 

#kwargs using
def val_assign(**kwargs):
    print(kwargs)#dictionary hisebe rakhbe
    for key,value in kwargs.items():
        print(f"Key {key} and values: {value}")

val_assign(a=10,b=20,c=40,n_estimator=50)
```

```python
#generator 
#create chunk to make memory more useful for big file
lst=[x for x in range(1000)]
print(lst)

def data_load(cn_size,lst):
    for i in range(0,len(lst),cn_size):
        yield lst[i:i+cn_size]

x=data_load(10,lst)
print(type(x))
print(x)
print(next(x))
print(next(x))

```
