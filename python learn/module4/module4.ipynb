```python
#5.1(Touples)
#Touples
#collection of data
# immutable 
```

```python
#declare touple
tup=(1,2,3,4,5)
print(type(tup),tup)
float_tup=(1.2,3.4,5.6)
print(type(float_tup),float_tup)
mixed_tup=(1,2.3,"hello",True)
print(type(mixed_tup),mixed_tup)
#one value can be declared as touple by adding comma after value
single_tup=(1,)
print(type(single_tup),single_tup)

#any list can be converted to touple by using tuple() function
my_list=[1,2,3,4,5]
tup_form=tuple(my_list)
print(type(tup_form),tup_form)

```

```python
#acessing touple values

tup1=(1,2,3,4,5)#its index start from 0
print(tup1[0])

#slicing in touples

new_tup=tup1[0:2]
print(new_tup)

#immutable
list=[10,20,30,40,50]
list.append(60)
list[1]=70
print(list)

#we can change value in list and can add value then the list change but touple dont change

#tup1[1]=20 #error will occur cause no value can assign in touple 

print(tup1)
```

```python
#touple methods only a few like count() ,index()
tup=(10,29,30,10,33,50,10,59)
print(tup.count(10))
print(tup.index())
```
