```python
#declare touple
tup=(1,2,3,4,5)
print(type(tup),tup)
float_tup=(1.2,3.4,5.6)
print(type(float_tup),float_tup)
mixed_tup=(1,2.3,"hello",True)
print(type(mixed_tup),mixed_tup)
#one value can be declared as touple by adding comma after value
single_tup=(1,)
print(type(single_tup),single_tup)

#any list can be converted to touple by using tuple() function
my_list=[1,2,3,4,5]
tup_form=tuple(my_list)
print(type(tup_form),tup_form)

```

```python
#acessing touple values

tup1=(1,2,3,4,5)#its index start from 0
print(tup1[0])

#slicing in touples

new_tup=tup1[0:2]
print(new_tup)

#immutable
list=[10,20,30,40,50]
list.append(60)
list[1]=70
print(list)

#we can change value in list and can add value then the list change but touple dont change

#tup1[1]=20 #error will occur cause no value can assign in touple 

print(tup1)
```

```python
#touple methods only a few like count() ,index()
tup=(10,29,30,10,33,50,10,59)
print(tup.count(10))
print(tup.index(33))
```

```python
#4.2 set in python
#Set
#we will learn set now
#unique element no duplicate in one set,mutable, unordered(indexing and slicing not possible)
set1={1,2,3}
print(type(set1),set1)

#empty set 
a=[]
print(type(a))#list
b=()
print(type(b))#touple
c={}
print(type(c))#set but will show dict
d=set()
print(type(d))# this will show an empty set
```

```python
#using set if else and forloop

set2={1,2,4,5,6,1,4}
print(set2)#all unique value will be shown 
if 3 in set2:
    print("3 is present in set")
else:
    print("Not present")

sum=0
for ele in set2:
    print(ele)
    sum+=ele

print(sum)
```

```python
#set methods
s={3,5,2,1,4,6}
print(s) #orderly print hbe though its unsorted

s.add(10)
s.add(0)
print(s)

s.pop()
print(s)

s.remove(3)
print(s)
```

```python
#union ,intersection,disjoint,subset
a={1,3,5,7,8,3}
b={2,3,5,4,7,10}

c={1,3,5}
print(a.union(b))#all unique values
print(a.intersection(b))#common
print(a.isdisjoint(b))#any common point
print(a.issubset(b))
print(c.issubset(a))

```

```python
#4.3 dictionary basic
#store value as (key,value),mutable means can change value,also unordered so indexing can apply
# key-> always immutable can be (int,float,tuple,string) 
#value-> any thing can store (string,int,float,set,touple,dictionary)

#empty dictionary
dic={}
print(type(dic))
#insert value
dic={"name":"shihab","age":22,"address:":"Dhaka","Number":[10,20,30]}
print(dic)

#access dic
#find through key -> dic_name[key]
print(dic["name"],dic["address:"])
print(dic.get("age"))

#assign value

dic["name"]="kuddus"
print(dic)

dic={"name":"shihab","age":22,"address:":"Dhaka","Number":[10,20,30],"age":34}#assign duplicate age
print(dic)#age will be the second one
```

```python
#4.4 Dictionary methods
#you can print any value for dictionary(print default value)
print(dic)
print(dic.get("name_of_raja"))#this will show none
print(dic.get("name_of_raja","sultan ali khan"))

#adding new key in dictionary

dic["math_marks"]=89
print(dic)

dic.update({"english_mark":40})
print(dic)

#delete any key
del dic["math_marks"]
print(dic)
#copy any dictonary to another 
dic_2=dic.copy()
print(dic_2)
#pop
dic_2.pop("name")
print(dic_2)

#dic_3={[10,20,30]:"a"}  #this will show error cause this is a list(TypeError: unhashable type: 'list')

#dic_3={{10,20,30}:"aa"} # (TypeError: unhashable type: 'set)
dic_3={(10,20,30):"aa"} # only tupple allow in key (cause tuple is hashable)
print(dic_3)




```

```python
#4.5 dictionary view methods
#keys and value separately generate
dic={"name":"shihab","age":22,"address:":"Dhaka","Number":[10,20,30]}

keys=dic.keys()
print(keys)

dic["Math_result"]=100
print(keys)

value=dic.values()
print(value)

items=dic.items()#ekta list korbe and list er vitor value gula tuple hisebe dekhabe
print(items)

#when there is many complex value in dictionary we use this item 
for keys,value in dic.items():
    print(keys,value)
```
