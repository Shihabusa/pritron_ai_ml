```python
#practice list[]
#allow duplicate value
t_list=["apple","banana","cherry","apple","cherry"]
print(t_list)
#find length
print(len(t_list))

#can store any data type
list1=[1,2,3,5,7,9]
list2=[True,False,True,True,False]

list3=["abc",34,True,40,"male"]#can store any data type in any variable
print(type(list3))

#construct new list
thislist = list(("apple", "banana", "cherry")) # note the double round-brackets
print(thislist)

#list & tuple allow duplicates members
#list & dictionary is ordered and changeable
#Tuple & set are unchangeable 
thislist = ["apple", "banana", "cherry"]
print(thislist[-1])
```

```python
#check item present in a list
the_list=["apple", "banana", "cherry", "orange", "kiwi", "mango"]
if "apple" in the_list:
    print("we found it")

the_list[1:3]=["blackberry","watermelon"]
print(the_list)
#change 1 value with 2 value
thislist = ["apple", "banana", "cherry"]
thislist[1:2] = ["blackcurrant", "watermelon"]
print(thislist)
#change 2 value with one value
thislist[1:3] = ["rose"]
print(thislist)

#insert()
thislist.insert(3,"banana")
print(thislist)

#append() add at the end
thislist.append("orange")
print(thislist)

#extend
thislist.extend(the_list)


```

```python
#remove item from a list
tlist=["apple", "banana", "cherry", "orange","banana", "kiwi", "mango"]
tlist.remove("banana")#remove the first one
print(tlist)
tlist.pop()#last element will be pop
print(tlist)
tlist.pop(2)#specific element will be pop
print(tlist)

#clear list
#del tlist
#print(tlist)#error asbe cause eta puro veriable delete kore dise 
tlist.clear()#list will be empty list
print(tlist)
```
