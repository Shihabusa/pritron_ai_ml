```python
#While loop 
count =1
while(count<=10):#until the condition is false ,the loop will run
    print(count,"Hello world")
    count+=1
```

```python
#continue statement and break statement
sum=0
for i in range(1,11):
    if( i%2 ==0):#when i is even number, we will skip the rest of the code and continue with next iteration
        continue
    
    sum+=i
    print(i,"is added to sum")
print(sum)

#break statement
accuracy =95
for i in range (20):
    accuracy+=1
    print(accuracy)
    if(accuracy==100):
        print(accuracy,"is achieved")
        break
print(accuracy,"is the final accuracy")
```

```python
#problem K (max min)
inp=input()
numbers=inp.split()
x=int(numbers[0])
y=int(numbers[1])
z=int(numbers[2])

max=x 
min=x

#for min
if y<min:
    min=y
if z<min:
    min=z

#for max
if y>max:
    max=y
if z>max:
    max=z

print( min, max)

```
