```python
#string indexing
name="Pritron"
print(name[0])  # Output: P

#string slicing
full_name="Pritron AI ML"
print(full_name[0:6:1])  # Output: Pritron here in [ start:end:step ]
print(full_name[0:6:2])  # Output: Pto
#BACKWORD INDEXING E LAST INDEX IS -1
print(full_name[-1])  # Output: L
```

```python
#STRING MEETHODS
#lower(),upper(),title(),capitalize(),count(),find(),replace(),split(),join()
string="Pritron AI ML"
print(string.lower()) #all letters in lower case
print(len(string)) #length of string
print(string.count("A")) #count of A in string
print(string.find("AI")) #find the index of "AI" in string
print(string.replace("ML", "Machine Learning")) #replace "ML" with "Machine Learning"
print(string.split(" ")) #split the string into a list of words
print(" ".join(["Pritron", "AI", "ML"])) #join the list of words into a string
print("Pritron" in string) #check if "pritron" is in string
```
