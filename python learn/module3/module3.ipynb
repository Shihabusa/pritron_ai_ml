```python
4.6 #list as stack and queue
numbers.sort()
print(numbers)

numbers.reverse()
print(numbers)

#stack as list
#stack follow last in first out(LIFO) principle

stack=[]
type(stack)
stack.append(1)
stack.append(2)
stack.append(3)
stack.append(4)
stack.append(5)
print(stack)
print(stack.pop())#getting the top element of stack
print(stack)
print(stack.pop())
print(stack)

```

```python
#queue as list
#queue follow first in first out(FIFO) principle
queue=[]
queue.append(1)
queue.append(2)
queue.append(3)
queue.append(4)
queue.append(5)#new element will be added at last
print(queue)
print(queue.pop(0))#getting the first element of queue
print(queue)
print(queue.pop(0))#queue korar somoy first element ta ber hbe tai pop(0) use kora hbe
print(queue)
print(f"top element of stack is {stack[-1]}")#getting the top element of stack
print(f"first element of queue is {queue[0]}")#getting the first element of queue
```
